# Word Lemmatization

```python
# https://www.nltk.org/

# Import NLTK module in Python
import nltk

print("\n=============STEMMING===================\n")

# "Stemming is the process of reducing inflection in words to their root forms
# such as mapping a group of words to the same stem even if the stem itself is
# not a valid word in the Language."

# Import Stemmer module
from nltk.stem import PorterStemmer

# Import Tokenizer
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize

ps = PorterStemmer()

# Documentation: https://www.nltk.org/_modules/nltk/stem/porter.html
# Stemming Algorithm: https://tartarus.org/martin/PorterStemmer/

# Idea: Eliminate morphological and inflexional endings from words in English

# Stemming: A word stem is part of a word.
stem_words = ["game", "gaming", "gamed", "games", "leaves", "cared", "children",
              "sportingly", "fairly", "saw", "loving", "boy's"]


for word in stem_words:
  print(ps.stem(word))

print("\n=============USING DIFFERENT ALGORITHM===================\n")

# Do the stemming
from nltk.stem.snowball import SnowballStemmer

ps = SnowballStemmer("english")

for word in stem_words:
  print(ps.stem(word))

print("\n=============LEMMATIZATION===================\n")

# "Lemmatization, unlike Stemming, reduces the inflected words properly ensuring
# that the root word belongs to the language. In Lemmatization root word is called
# Lemma. A lemma (plural lemmas or lemmata) is the canonical form, dictionary
# form, or citation form of a set of words."

# Lemmatizer using WordNet's built-in morphy function. Returns the input word
# unchanged if it cannot be found in WordNet.

# Download NLTK wordnet
nltk.download('wordnet')

nltk.download('omw-1.4')  #Open Multilingual Wordnet
# Installing wordnet without OMW throws an error. It is caused by the
# introduction of a "provenances" dictionary (line 1176) to handle the new
# omw-1.4 directory structure. Building this dictionary requires the presence
# of the omw-1.4 package, which was not needed in the past.

lemma = nltk.wordnet.WordNetLemmatizer()

print(nltk.stem.WordNetLemmatizer().lemmatize('loving', 'v'))
print()

for word in stem_words:
  print(lemma.lemmatize(word))

print("\n======================================\n")

#Exercise: Use the best stemmer on below list of words and report your results.
# stem_words = [ "alumnus", "running", "alumnu", "alumni", "alumna", "alumnae", "cats", "catty", "cat", "chocolates", "chocolatey", "choco", "children" ]
```

# POS Tagging

```python
import nltk

# The Default Tagger

# The simplest possible tagger assigns the same tag to each token. This may
# seem to be a rather banal step, but it establishes an important baseline for
# tagger performance. In order to get the best result, we tag each word with
# the most likely word. (This kind of tagger is known as a majority class classifier).

# Question: What is the most frequency word category in a text?

# Now we can create a tagger, called default_tagger, that tags everything as NN.

# Sentence
data = "Well, it has been snowing throughout the day in Shimla."

# Tokenizer
nltk.download('punkt')
from nltk.tokenize import word_tokenize

# Returns a list of tokenized words
tokens = word_tokenize(data)

#POS tagger
default_tagger = nltk.DefaultTagger('NN')

# Print tagged words
print(default_tagger.tag(tokens))

print("\n=============USING DIFFERENT ALGORITHM===================\n")

# Regular expression tagger
# The regular expression tagger assigns tags to tokens on the basis of
# matching patterns. For instance, we might guess that any word ending in ed is
# the past participle of a verb, and any word ending with 's is a possessive noun.
# We can express these as a list of regular expressions:

patterns = [
  (r'.*ing$', 'VBG'),  # gerunds
  (r'.*ed$', 'VBD'),  # simple past
  (r'.*es$', 'VBZ'),  # 3rd singular present
  (r'.*\.$', 'SYM'),  # Punctuation
  (r'.*\,$', 'SYM'),  # Punctuation
  (r'.*ould$', 'MD'),  # modals
  (r".*\\'s$", 'NN$'),  # possessive nouns
  (r'.*s$', 'NNS'),  # plural nouns
  (r'^-?[0-9]+(.[0-9]+)?$', 'CD'),  # cardinal numbers
  (r'.*', 'NN')  # nouns (default)
]

# print(patterns)

regexp_tagger = nltk.RegexpTagger(patterns)
print(regexp_tagger.tag(tokens))

print("\n=============USING DIFFERENT ALGORITHM===================\n")

# An off-the-shelf tagger is available for English. It uses the Penn Treebank tagset.

# Maximum Entropy Tagger
nltk.download('averaged_perceptron_tagger_eng')
from nltk.tag import pos_tag

# Print all the tagged words
print(pos_tag(tokens))

#Documentation of tagset
# nltk.download('tagsets')
# print(nltk.help.upenn_tagset('RB'))
# print(nltk.help.upenn_tagset('NN.*'))

# Universal tagset
# Mapping schema: https://www.nltk.org/_modules/nltk/tag/mapping.html

nltk.download('universal_tagset')
print(pos_tag(tokens, tagset='universal'))

# Reading

# Mappings of treebank specific tagsets to a set of 12 universal part-of-speech tags.
# The 12 universal tags are:

# VERB - verbs (all tenses and modes)
# NOUN - nouns (common and proper)
# PRON - pronouns
# ADJ - adjectives
# ADV - adverbs
# ADP - adpositions (prepositions and postpositions)
# CONJ - conjunctions
# DET - determiners
# NUM - cardinal numbers
# PRT - particles or other function words
# X - other: foreign words, typos, abbreviations
# . - punctuation

# See "A Universal Part-of-Speech Tagset"
# by Slav Petrov, Dipanjan Das and Ryan McDonald
# for more details:
# http://arxiv.org/abs/1104.2086
```

# Frequency Distribution

```python
import nltk
from nltk.probability import FreqDist

# Concepts of dictionary will be useful here.

# One of the most useful thing to do on corpora is to count things:
# count words, count words with specific properties, count sequences, count tags.

# NLTK defines a useful class for this: nltk.FreqDist.
# (There is a closely related class in standard Python called Counter)

# FreqDist (Frequency Distribution) is an object that counts occurrences of objects.
# It is defined in the nltk.probability module.

tokens = ['a', 'b', 'a', 'a', 'c', 'b', 'a']
fdist = FreqDist(tokens)
print(fdist)

# Documentation: https://www.nltk.org/api/nltk.probability.html

print("a's count: " + str(fdist['a']))  # We saw 'a' 4 times
print("b's count: " + str(fdist['b']))  # We saw 'b' 2 time
print("c's count: " + str(fdist['c']))  # We saw 'c' 1 times
print("d's count: " + str(fdist['d']))  # We saw 'c' 0 times

print("max item: " + str(fdist.max()))  # 'a' is the most frequent sample seen
print("item type count: " +
      str(len(fdist)))  # we observed 2 types of objects in the stream
print("item type: " + str(fdist.keys())
      )  # Return the types of the objects that were observed in the stream
print("item count: " + str(fdist.values())
      )  # Return the types of the objects that were observed in the stream
print("a's probability: " +
      str(fdist.freq('a')))  # 2/3 of the samples we saw were 'a'
print("size of list with all items: " +
      str(fdist.N()))  # How many samples did we count?
print("size of list with unique items: " +
      str(fdist.B()))  # How many unique samples did we count?

print("\n===================Counting on a sentence======================\n")

# nltk.download('punkt')
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

sent = 'This is an example sentence to tell you how this example works'
tokens = word_tokenize(sent)
print(tokens)

fdist = FreqDist(tokens)
print(fdist)

print("item type: " + str(fdist.keys())
      )  # Return the types of the objects that were observed in the stream
print("item value: " + str(fdist.values())
      )  # Return the counts of the objects that were observed in the stream

# An equivalent way to do this is using list comprehension:

# TODO Solve it without list comprehension:

lower_tokens = [word.lower() for word in word_tokenize(sent)]
fdist = FreqDist(lower_tokens)
print(fdist)

print("item type: " + str(fdist.keys())
      )  # Return the types of the objects that were observed in the stream
```

# Corpus Counting

```python
import nltk

from nltk.probability import FreqDist

# Working on the Brown Corpus with NLTK

# NLTK contains a collection of tagged corpora, arranged as convenient Python objects. We will use the Brown corpus in this experiment. The tagged_sents version of the corpus is a list of sentences. Each sentence is a list of pairs (tuples) (word, tag). Similarly, one can access the corpus as a flat list of tagged words.

nltk.download('brown')
nltk.download('universal_tagset')

# Brown corpus
from nltk.corpus import brown

# Corpus documentation
# print(nltk.corpus.brown.readme())

print("\n==============WORD LEVEL=====================\n")

# Read POS tagged brown corpus
print((nltk.corpus.brown.tagged_words()))

# Reading POS tagged brown corpus with UNIVERSAL tagset
print(nltk.corpus.brown.tagged_words(tagset='universal'))

print("\n==============SENTENCE LEVEL=====================\n")

print(nltk.corpus.brown.tagged_sents()[0])
print('\n')

print(nltk.corpus.brown.tagged_sents(tagset='universal')[0])

print("\n============FREQUENCY DISTRIBUTION ON WORDS=================\n")

# Let us now check: which word is the most common among the about 100,000 words in this part of the Brown corpus? Which tag is the most common?

# Let's first make a list of just the words:

words = []
for (word, tag) in nltk.corpus.brown.tagged_words():
  words.append(word)

# An alternative way to do the exact same is this:
words = [word for (word, tag) in nltk.corpus.brown.tagged_words()]

# Next we'll count how often each word appears:
fdist_word = FreqDist(words)
print(fdist_word)

print("Total tokens: " +
      str(fdist_word.N()))  # We saw 1161192 words (1M) in the corpus

print("Word type: " +
      str(len(fdist_word)))  # How many distinct words are there?

print("Most frequent word: " +
      str(fdist_word.max()))  # What is the most frequent word?

print("Frequency of 'the': " +
      str(fdist_word['the']))  # How often does 'the' occur in the corpus

print("% count of 'the': " + str('%0.2f%%' % (fdist_word.freq('the') * 100)))

# We observe that the distribution of word tokens to word types is extremely
# unbalanced - a single word (the) accounts for over 5% of the word tokens.
# This is a general observation of linguistic data -- known as Zipf's Law --
# few types are extremely frequent, and many types are extremely rare.
```

### Exercise 1a: comments on the assigned tags

I judged each tag against the Penn Treebank conventions. Out of 54 tokens, **53 are tagged correctly and 1 is wrong**.

| # | Sentence | Verdict |
|---|----------|---------|
| 1 | Colorless green ideas sleep furiously. | **One mistake.** `Colorless` is tagged `NNP` (proper noun), but it is an adjective (`JJ`). The rest is right: `green/JJ ideas/NNS sleep/VBP furiously/RB`. |
| 2 | A bright, colorful rainbow appeared today. | All correct. `today` as `NN` is the Penn Treebank convention. |
| 3 | A villager went to city to sell his property along with his wife. | All correct. Both `to` get `TO` (infinitive marker and preposition share the tag in PTB), `his` is `PRP$`, `along with` is `IN IN`. |
| 4 | I saw the man with a telescope. | All correct. `saw` is `VBD`, not the noun/tool sense, because it follows the pronoun `I`. |
| 5 | We will sell gasoline to anyone in a glass container. | All correct. `anyone` is `NN` in PTB and the noun modifier `glass` is `NN`. |
| 6 | Get the cat with the gloves. | All correct. Imperative `Get` is `VB`, `gloves` is `NNS`. |

**What goes wrong, and why**

1. **Unknown word plus capital letter (sentence 1).** `Colorless` is rare, and it starts the sentence with a capital letter. The tagger cannot tell a sentence-initial capital from a proper-noun capital, so it guesses `NNP`. The suffix `-less` (a strong adjective cue) was not enough to override that. A *rare word* problem, not a grammar problem.
2. **Meaningless but grammatical sentence (sentence 1).** The tagger only uses word form and neighbouring tags, never meaning, so the nonsense does not confuse it. It still tags the rest correctly.
3. **Structural ambiguity is invisible to a POS tagger (sentences 4, 5, 6).** "with a telescope" can attach to *saw* or to *the man*, "in a glass container" to *sell* or to *gasoline*, and "with the gloves" to *Get* or to *the cat*. Both readings use exactly the same POS tags, so the ambiguity is a *parsing* (attachment) problem, not a tagging mistake.

```python
# Find the top 10 most frequent POS tags using brown corpus!

import nltk
from nltk.probability import FreqDist

nltk.download('brown')
nltk.download('universal_tagset')
from nltk.corpus import brown

# a) Brown's own tagset
tags = []
for (word, tag) in brown.tagged_words():
  tags.append(tag)

fdist_tag = FreqDist(tags)
print("Total tag tokens :", fdist_tag.N())
print("Distinct tags    :", fdist_tag.B())
print("\nTop 10 tags (Brown tagset)")
for rank, (tag, count) in enumerate(fdist_tag.most_common(10), start=1):
  print("%2d. %-5s %7d  (%.2f%%)" % (rank, tag, count, 100.0 * count / fdist_tag.N()))

# b) Universal tagset (12 coarse tags)
utags = []
for (word, tag) in brown.tagged_words(tagset='universal'):
  utags.append(tag)

fdist_utag = FreqDist(utags)
print("\nTop 10 tags (Universal tagset)")
for rank, (tag, count) in enumerate(fdist_utag.most_common(10), start=1):
  print("%2d. %-5s %7d  (%.2f%%)" % (rank, tag, count, 100.0 * count / fdist_utag.N()))
```

**Observation (1b).** With Brown's own tagset the most frequent tag is `NN` (singular common noun), followed by `IN` (preposition/subordinating conjunction) and `AT` (article). Punctuation (`.` and `,`) also makes the top 10. With the universal tagset, `NOUN` comes first, then `VERB`, then punctuation `.`. Tags such as `NN`, `NNS` and `NP` are merged into one `NOUN` class, so the universal counts are much larger per tag.
